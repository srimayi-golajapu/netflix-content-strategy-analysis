# Exploratory Data Analysis (EDA)

## Business Question 1:
### Has Netflix's content library grown over time?

```python
df["year_added"]=df["date_added"].dt.year
```

```python
titles_per_year= df["year_added"].value_counts().sort_index()
```

```python
titles_per_year
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))
plt.plot(
titles_per_year.index,
titles_per_year.values,
marker='o')
plt.title("Number of titles added to Netflix each year")
plt.xlabel("Year")
plt.ylabel("Number of Titles added")
plt.grid(True)
plt.show()
```

# Observation:
## From 2008 to 2014, Netflix added relatively few titles each year. Starting in 2015, the number of titles added increased rapidly, reaching its peak in 2019. After 2019, the number of titles added declined in 2020 and 2021.

# Insight:
## The rapid increase in content additions after 2015 suggests that Netflix significantly expanded its content acquisition and production strategy. The decline after 2019 may indicate a shift toward focusing on quality, operational constraints, or other strategic factors, but the dataset alone cannot determine the exact reason.

# Buisness Implication:
## A larger content library increases the variety of content available to viewers. This strategy can help Netflix appeal to a broader audience and strengthen its competitive position in the streaming market.

## Business Question 2

### Has Netflix shifted its focus from Movies to TV Shows over time?

```python
content_by_type = (df.groupby(["year_added", "type"]).size().unstack())
```

```python
content_by_type
```

```python
content_by_type=content_by_type.fillna(0)
```

```python
content_by_type
```

```python
plt.figure(figsize=(10,5))
plt.plot(content_by_type.index, content_by_type["Movie"], marker="o", label="Movie")
plt.plot(content_by_type.index, content_by_type["TV Show"], marker="o", label="TV Shows")
plt.title("Movie andTV Shows Added over time")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")
plt.xticks(content_by_type.index, rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

# Observation
## From 2013 onward, both Movies and TV Shows showed an upward trend in the number of titles added. However, the number of Movies grew much faster than TV Shows, especially between 2015 and 2019. Both categories reached their highest additions around 2019 and declined afterwards.

# Insights
## Netflix expanded both Movies and TV Shows over time, but Movies remained the dominant content type throughout the observed period. Rather than replacing Movies with TV Shows, Netflix appears to have expanded both categories while maintaining a stronger focus on Movies.

# Business Implication
## Netflix has consistently expanded both Movies and TV Shows over time while maintaining Movies as the dominant content type. Continuing to invest in both categories can help serve diverse audience preferences and maintain a balanced content library.

# Observation
## The United States has the highest number of titles in the Netflix catalog, contributing significantly more content than any other country.
## India is the second-largest contributor, followed by the United Kingdom.
## After the top three countries, the number of titles drops considerably, with countries such as Canada, France, Japan, Spain, South Korea, Germany, and Mexico contributing relatively smaller amounts.

# Insight
## Netflix's content library is heavily concentrated in a few countries, particularly the United States, which dominates the platform's catalog.
## While Netflix offers content from multiple regions, the majority of titles come from a limited number of countries with well-established entertainment industries.
## The presence of countries from North America, Europe, and Asia among the top contributors reflects Netflix's broad international content acquisition strategy.

# Business Implication
## Netflix should continue investing in major content-producing markets like the United States and India, as they represent a substantial portion of the platform's catalog.
## To strengthen its global reach, Netflix could expand investments and partnerships in underrepresented regions, helping diversify its content library and attract audiences from emerging markets.
## A more geographically balanced catalog can improve regional subscriber engagement and reduce reliance on a few dominant content markets.

# Business Question 5
## Does Netflix focus more on family-friendly content or mature (adult) content?

```python
family = ["TV-Y", "TV-Y7", "TV-G", "G", "PG", "TV-PG"]
df["content_category"] = "Mature"
df.loc[df["rating"].isin(family), "content_category"] = "Family-Friendly"
content_category = df["content_category"].value_counts()
```

```python
content_category
```

```python
plt.figure(figsize=(6,6))
content_category.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Family-Friendly and Mature Content on Netflix")
plt.ylabel("")
plt.show()
```

# Observation
## The pie chart shows that mature content accounts for a significantly larger share of Netflix's catalog, while family-friendly content represents a smaller proportion. This indicates that a majority of the titles available on Netflix are intended for teenage and adult audiences rather than young children or families.

# Insight
## The analysis suggests that Netflix primarily focuses on providing content for mature audiences, including teenagers and adults. Although the platform offers a selection of family-friendly titles, its content strategy appears to be driven by the higher demand for movies and TV shows with mature themes, diverse storytelling, and broader audience appeal.

# Business Implication
## The findings indicate that Netflix should continue investing in mature content, as it forms the largest portion of its catalog and aligns with its primary target audience. At the same time, expanding high-quality family-friendly content could help attract more households with children, broaden the subscriber base, and strengthen customer retention by catering to viewers across different age groups.

The distribution shows that a small number of directors have multiple titles on the platform.

# Insight

## Netflix collaborates repeatedly with certain directors, indicating long-term partnerships or frequent licensing of their work. At the same time, the presence of many different directors in the top rankings suggests that Netflix maintains a diverse network of creators rather than relying on a single filmmaker.

# Business Implication

## Netflix can strengthen relationships with high-performing directors who consistently contribute popular content while continuing to onboard new directors. Balancing recurring collaborations with fresh creative talent helps maintain a diverse and engaging content library.

## Netflix's catalog includes recurring collaborations with well-known actors, particularly from the Indian film industry. Featuring popular actors helps attract audiences, increases content discoverability, and strengthens Netflix's appeal in regions where these actors have a large fan base.

# Business Implication

## Netflix can continue acquiring or producing content featuring highly recognized actors, as they can attract and retain viewers. At the same time, introducing emerging talent alongside established actors can help diversify the content library while maintaining audience engagement.
